```python
import pandas as pd
```

```python
df = pd.read_csv(r"C:\Users\HP\Desktop\SQL\archive\sales_data_sample.csv", encoding='latin1')
df
```

```python
df = df.drop_duplicates()
df
```

```python
df = df.drop(columns = "ADDRESSLINE2")
df
```

```python
df = df.drop(columns = "STATE")
df
```

```python
df = df.drop(columns = "TERRITORY")
df
```

```python
df.rename(columns={ 'ORDERNUMBER': 'Order_Number'})
```

```python
df = df.rename(columns={ 'QUANTITYORDERED': 'Quantity_Ordered'})
df
```

```python
df = df.rename(columns={ 'PRICEEACH': 'Price_Each'})
df
```

```python
df = df.rename(columns={ 'ORDERLINENUMBER': 'Order_Line_Number',
                   'ORDERDATE': 'Order_Date',
                   'SALES': 'Sales',
                   'STATUS': 'Status',
                   'PRODUCTCODE': 'Product_Code',
                   'PHONE	': 'Phone',
                   'CITY': 'City',
                   'CONTACTLASTNAME': 'Last_Name',
                   'CONTACTFIRSTNAME': 'First_Name',
                   'DEALSIZE': 'Deal_Size'
})

df
```

```python
df = df.rename(columns={ 'MONTH_ID': 'Month_ID',
                   'YEAR_ID	': 'Year_ID	',
                   'CUSTOMERNAME': 'Company_Name',
                   'PHONE': 'Phone',
                   'ADDRESSLINE1': 'Address',
                   'POSTALCODE	': 'Postal_Code',
                   'COUNTRY': 'Country'
})

df
```

```python
df["Last_Name"] = df['Last_Name'].str.strip()
df
```

```python
df['First_Name'] = df['First_Name'].str.strip()
df
```

```python
df["Phone"] = df["Phone"].str.replace(r'[^0-9]'  , '', regex=True)
df
```

```python
df["POSTALCODE"] = df["POSTALCODE"].fillna(0)
df
```

```python
df.drop_duplicates(subset= ['Address'])
df
```

```python
df= df.drop(columns=['Order_Line_Number', 'Month_ID', 'YEAR_ID' ])
df
```

```python
df = df.rename(columns={ 'ORDERNUMBER': 'Order_Number'})
df
```

```python
df['Order_Date'] = pd.to_datetime(df['Order_Date']).dt.date
df['Order_Date']
```

```python
df
```

```python
df = df.drop(columns=['Phone', 'Address','POSTALCODE','Last_Name', 'First_Name', 'MSRP', 'Product_Code'])
df
```

```python
df
```

```python
print(df.columns.tolist())
```

```python
##  Month with the highest sales

monthly_sales = df.groupby('QTR_ID')['Sales'].sum().sort_values(ascending=False)
print(monthly_sales)
```

```python
##  The country that made the most revenue

country_sales = df.groupby('Country')['Sales'].sum().sort_values(ascending=False)
print(country_sales)
```

```python
## Product line which sold the most

product_sales = df.groupby('PRODUCTLINE')['Sales'].sum().sort_values(ascending=False)
print(product_sales)
```

```python
## Top 10 customers

top_customers = df.groupby('Company_Name')['Sales'].sum().sort_values(ascending=False).head(10)
print(top_customers)
```

```python
## Which deal size brings most revenue

deal_sales = df.groupby('Deal_Size')['Sales'].sum().sort_values(ascending=False)
print(deal_sales)
```

```python
##  Best performing quarter

quarter_sales = df.groupby('QTR_ID')['Sales'].sum().sort_values(ascending=False)
print(quarter_sales)
```

```python
## Total overall revenue

total_revenue = df['Sales'].sum()
print(f'Total Revenue: ${total_revenue:,.2f}')
```

```python
## Total number of orders

total_orders = df['Order_Number'].count()
print(f'Total Orders: {total_orders}')
```

```python
## Average sales per order

avg_sales = df['Sales'].mean()
print(f'Average Sales Per Order: ${avg_sales:,.2f}')
```

```python
## Best performing city

city_sales = df.groupby('City')['Sales'].sum().sort_values(ascending=False).head(10)
print(city_sales)
```

```python
## Order status breakdown

status_count = df['Status'].value_counts()
print(status_count)
```
