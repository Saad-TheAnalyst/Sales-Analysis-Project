```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

df = pd.read_csv('sales_data_sample.csv', encoding='unicode_escape')
df['ORDERDATE'] = pd.to_datetime(df['ORDERDATE'])
df['MONTH_ID'] = df['ORDERDATE'].dt.month
df['YEAR_ID'] = df['ORDERDATE'].dt.year
df['TotalRevenue'] = df['QUANTITYORDERED'] * df['PRICEEACH']
print('Data loaded successfully')
print(df.shape)
```

```python
##   Monthly Sales Trend

monthly_sales = df.groupby(['YEAR_ID', 'MONTH_ID'])['TotalRevenue'].sum().reset_index()
print("=== MONTHLY SALES TREND ===")
print(monthly_sales.to_string())
```

```python
##   Year Over Year Growth

yearly_sales = df.groupby('YEAR_ID')['TotalRevenue'].sum()
print("=== YEAR OVER YEAR SALES ===")
print(yearly_sales)
yoy_growth = yearly_sales.pct_change() * 100
print("\nYear Over Year Growth:")
print(yoy_growth)
```

```python
##   Product Line Profitability

product_profit = df.groupby('PRODUCTLINE').agg(
    Total_Revenue=('TotalRevenue', 'sum'),
    Total_Quantity=('QUANTITYORDERED', 'sum'),
    Avg_Price=('PRICEEACH', 'mean')
).round(2)
product_profit['Revenue_Per_Unit'] = (product_profit['Total_Revenue'] / product_profit['Total_Quantity']).round(2)
print("=== PRODUCT LINE PROFITABILITY ===")
print(product_profit.sort_values('Revenue_Per_Unit', ascending=False).to_string())
```

```python
##  Customer Segmentation

customer_revenue = df.groupby('CUSTOMERNAME')['TotalRevenue'].sum()
customer_segments = pd.cut(customer_revenue,
    bins=[0, 50000, 150000, float('inf')],
    labels=['Small', 'Medium', 'Large'])
segment_counts = customer_segments.value_counts()
print("=== CUSTOMER SEGMENTATION ===")
print(segment_counts)
print(f"\nTotal Customers: {len(customer_revenue)}")
```

```python
##  Top 10 Cities by Revenue

city_revenue = df.groupby('CITY')['TotalRevenue'].sum().sort_values(ascending=False).head(10)
print("=== TOP 10 CITIES BY REVENUE ===")
print(city_revenue)
```

```python
##  Order Status Analysis

status_analysis = df.groupby('STATUS')['TotalRevenue'].agg(['count', 'sum']).round(2)
status_analysis.columns = ['Order Count', 'Total Revenue']
print("=== ORDER STATUS ANALYSIS ===")
print(status_analysis.sort_values('Order Count', ascending=False).to_string())
```

```python
##  Deal Size vs Quantity

deal_quantity = df.groupby('DEALSIZE').agg(
    Total_Orders=('ORDERNUMBER', 'count'),
    Total_Quantity=('QUANTITYORDERED', 'sum'),
    Total_Revenue=('TotalRevenue', 'sum'),
    Avg_Order_Value=('TotalRevenue', 'mean')
).round(2)
print("=== DEAL SIZE ANALYSIS ===")
print(deal_quantity.to_string())
```

```python
## Visualization
```

```python
##  Monthly Sales Trend

plt.figure(figsize=(14,6))
for year in [2003, 2004, 2005]:
    data = monthly_sales[monthly_sales['YEAR_ID']==year]
    plt.plot(data['MONTH_ID'], data['TotalRevenue'], marker='o', linewidth=2, label=str(year))
plt.title('Monthly Sales Trend by Year', fontsize=14, fontweight='bold')
plt.xlabel('Month')
plt.ylabel('Total Revenue ($)')
plt.xticks(range(1,13), ['Jan','Feb','Mar','Apr','May','Jun','Jul','Aug','Sep','Oct','Nov','Dec'])
plt.legend()
plt.tight_layout()
plt.show()
```

```python
##  Year Over Year Growth

fig, ax1 = plt.subplots(figsize=(10,6))
yearly_sales.plot(kind='bar', ax=ax1, color='steelblue', alpha=0.7)
ax1.set_title('Year Over Year Sales Performance', fontsize=14, fontweight='bold')
ax1.set_xlabel('Year')
ax1.set_ylabel('Total Revenue ($)')
plt.xticks(rotation=0)
plt.tight_layout()
plt.show()
```

```python
##  Product Line Profitability

plt.figure(figsize=(12,6))
product_profit['Revenue_Per_Unit'].sort_values().plot(kind='barh', color='orange')
plt.title('Product Line Revenue Per Unit Sold', fontsize=14, fontweight='bold')
plt.xlabel('Revenue Per Unit ($)')
plt.tight_layout()
plt.show()
```

```python
##  Customer Segmentation

plt.figure(figsize=(8,6))
segment_counts.plot(kind='pie', autopct='%1.1f%%', colors=['steelblue', 'orange', 'green'])
plt.title('Customer Segmentation by Revenue', fontsize=14, fontweight='bold')
plt.ylabel('')
plt.tight_layout()
plt.show()
```

```python
##  Top 10 Cities by Revenue

plt.figure(figsize=(12,6))
city_revenue.sort_values().plot(kind='barh', color='purple')
plt.title('Top 10 Cities by Revenue', fontsize=14, fontweight='bold')
plt.xlabel('Total Revenue ($)')
plt.tight_layout()
plt.show()
```

```python
##  Order Status Analysis

plt.figure(figsize=(10,6))
status_analysis['Order Count'].sort_values().plot(kind='barh', color='steelblue')
plt.title('Order Status Breakdown', fontsize=14, fontweight='bold')
plt.xlabel('Number of Orders')
plt.tight_layout()
plt.show()
```

```python
##   Deal Size Analysis

fig, axes = plt.subplots(1, 2, figsize=(14,6))
deal_quantity['Total_Revenue'].plot(kind='bar', ax=axes[0], color='steelblue')
axes[0].set_title('Total Revenue by Deal Size', fontweight='bold')
axes[0].set_xlabel('Deal Size')
axes[0].set_ylabel('Total Revenue ($)')
axes[0].tick_params(axis='x', rotation=0)

deal_quantity['Avg_Order_Value'].plot(kind='bar', ax=axes[1], color='orange')
axes[1].set_title('Average Order Value by Deal Size', fontweight='bold')
axes[1].set_xlabel('Deal Size')
axes[1].set_ylabel('Average Order Value ($)')
axes[1].tick_params(axis='x', rotation=0)
plt.tight_layout()
plt.show()
```
